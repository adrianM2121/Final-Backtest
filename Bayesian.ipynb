```python
import numpy as np




#Fair Coin: 

Prior_H1 = 0.5 #Prior hypothesis for coin 1
Prior_H2 = 0.5 #Prior hypothesis: coin is biased

#chosen H1 to be the fair coin and H2 to be the biased coin with 0.7 / 0.3 probability of Heads and Tails

def likelihood(data, hypothesis):
    if hypothesis == "H1":
        return 0.5
    elif hypothesis == "H2":
        return 0.5 #if data == 'H' else 0.3

def bayesian_update(prior_H1, prior_H2, data):

    likelihood_H1 = likelihood(data, "H1")
    likelihood_H2 = likelihood(data, "H2")


    #total evidence:
    P_data = likelihood_H1 * prior_H1 + likelihood_H2 * prior_H2

    Posterior_H1 = (likelihood_H1 * prior_H1) / P_data
    Posterior_H2 = (likelihood_H2 * prior_H2) / P_data

    return Posterior_H1, Posterior_H2

amount_of_flips = np.random.choice([1,0], size = 20) #randomly choose between 1 and 0 in array of 20

for flip in amount_of_flips:
    Prior_H1, Prior_H2 = bayesian_update(Prior_H1, Prior_H2, flip)
    
    print(f"after observing {flip}: P(H1 | Data) = {Prior_H1:.4f}, P(H2 | Data) = {Prior_H2:.4f}")

```
