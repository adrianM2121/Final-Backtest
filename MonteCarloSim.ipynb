```python
import numpy as np
import matplotlib.pyplot as plt




#estimating pi by throwing a dart inbetween a square from (-1,1) to (1,1) with a circle of area pi inside the square. 


rng = np.random.default_rng(seed=42)
N = 100000 #number of sim ran

x = rng.uniform( -1, 1, N) #setting x values of the square
y = rng.uniform(-1,1,N) #setting y v
inside = (x**2 + y**2) <= 1 #** means exponential so it would be x^2 + y^2 <=1 which is the equation of the inside circle with radius 1

piEstimate = 4*inside.mean() #since the ratio the darts hits is ratio of areas (pi/4), then estimating pi should be 4 pi
meanInside = inside.mean()
standardError = np.sqrt(meanInside*(1-meanInside)/N) 

print(f"Estimage of pi: {piEstimate:.5f}") # :.2f is for formatting the float to 2 decimal places
print(f"Standard error: {standardError:.5f}")







#graphing

n_show = 5000

xs, ys, ins = x[:n_show], y[:n_show], inside[:n_show]

fig, ax = plt.subplots(figsize=(6, 6))

ax.scatter(xs[ins],  ys[ins],  s=3, color="#2f6f8f", alpha=.5, linewidths=0, label="inside")
ax.scatter(xs[~ins], ys[~ins], s=3, color="#c05621", alpha=.5, linewidths=0, label="outside")

theta = np.linspace(0, 2*np.pi, 400)
ax.plot(np.cos(theta), np.sin(theta), color="black", lw=1.5)

ax.set_aspect("equal")
ax.set_xlabel("x"); ax.set_ylabel("y")
#ax.set_title(f"first {n_show:,} of {N:,} darts\n$\\pi \\approx$ {piEstimate:.5f} ± {standardError:.5f}")
ax.legend(markerscale=4, loc="upper right")

fig.tight_layout()
plt.show()
```
