```python
import numpy as np
import matplotlib.pyplot as plt




#FORMULAS: 

def set_frame(xInitial, xFinal, num):
    plt.figure()                                  # start a fresh figure
    x=np.linspace(xInitial, xFinal, num)         #3setting frame of axis, num for how many times linspace generates 
    y=np.zeros(xFinal)                      #set y values
    y[0]=xInitial                           #initial value of y
    return x

def make_graph_and_mean(iterations_Amount, xFinal, delta_t, xInitial, x):
    mean_Final = 0
    j=0
    for j in range(0, iterations_Amount):
        y = np.zeros(xFinal)
        y[0] = xInitial

        for i in range(delta_t, int(xFinal), delta_t):   
            middle_Step = 0
            y[i]=y[i-delta_t]*np.exp((mean-0.5*standardDeviation**2)*delta_t + standardDeviation*(np.sqrt(delta_t)*np.random.normal(0,1)))
            middle_Step += y[i] 

    
        plt.plot(x[::delta_t], y[::delta_t])
        mean_Final += middle_Step 
    mean_Final = mean_Final / iterations_Amount
    plt.show()
    return mean_Final


#SPECIFICATIONS: 

mean=0                      #the drfit direction
standardDeviation=0.001     #the spread or volatility
xInitial = 1                #start of graph
xFinal = 100                #end of graph   / also number of points generated (in third slot of set_frame)
iterations_Amount = 100     #number of individual paths
delta_t = 1              #( delta_t / total spaces ) = number of times formula is applied

num_steps = xFinal/delta_t


#Note: to change the amount of times the formula is process, change the xFinal and num (should be equal), then change delta accordingly. Delta determines the count. 




#MAIN: 
x = set_frame(xInitial, xFinal, xFinal) 


print(f"mean: {make_graph_and_mean(iterations_Amount, xFinal, delta_t, xInitial, x)}")












```
